```python
years = [2014, 2015, 2016, 2017,
         2018, 2019, 2020, 2021]

income = [55, 56, 62, 61,
          72, 72, 73, 75]

income_ticks = list(range(50, 81, 2))
```

```python
plt.plot(years, income)
plt.title("Income of john")
```

```python
plt.plot(years, income)
plt.title("Income over the years") # fontsize = "", # fontname = " "
plt.xlabel("Years")
plt.ylabel("Income in thousands")
plt.yticks(income_ticks, [f"{x}k USD" for x in income_ticks])
plt.show()

```
